# 7. Criar cópia para análise

```python
df_copy = df.copy()
```

```python
# mapear estado Civil
mapeamento_ec = {1: 'Casado/União Estável', 2: 'Divorciado', 3: 'Separado', 4: 'Solteiro', 5: 'Viúvo'}
df_copy['CL_EC_TEXTO'] = df_copy['CL_EC'].map(mapeamento_ec)
```

```python
df_copy
```

```python
# verificar estado civil
print("QUANTIDADE DE REGISTROS POR ESTADO CIVIL:")
(df_copy['CL_EC_TEXTO'].value_counts())
```

```python
 # mostra clientes com IDs diferentes
clientes_unicos = df_copy.drop_duplicates(subset=['CL_ID'])
```

```python
clientes_unicos
```

```python
# analisar gêneros

print("QUANTIDADE DE CLIENTES CADASTRADOS POR GÊNERO:")
(clientes_unicos['CL_GENERO'].value_counts())
```

```python
# agrupamento de gênero e total de compras
compras_genero = df.groupby('CL_GENERO').size().reset_index(name='TOTAL_COMPRAS')
(compras_genero)
```

```python
compras_unicas = df.drop_duplicates(subset=['CO_ID'])
total_de_compras = len(compras_unicas)
```

```python
# Compras (total) por gênero
resultado = compras_unicas.groupby('CL_GENERO').size().reset_index(name='TOTAL_COMPRAS')
(resultado)
```

```python
# total de filhos por gênero

total_filhos_genero = compras_unicas.groupby('CL_GENERO')['CL_FHL'].sum().reset_index(name='TOTAL_FILHOS')
(total_filhos_genero)
```

```python
# média de filhos
filhos_por_genero = compras_unicas.groupby('CL_GENERO')['CL_FHL'].mean().reset_index(name='MEDIA_FILHOS')
(filhos_por_genero)
```

```python
# Agrupamenteo de gênero e desvio padrão da coluna de filhos
desvio_por_genero = clientes_unicos.groupby('CL_GENERO')['CL_FHL'].std().reset_index(name='DESVIO_FILHOS')
(desvio_por_genero)
```

```python
# Categoria do produto mais vendida

categoria_vendas = df.groupby('PR_CAT').size().reset_index(name='QUANTIDADE_VENDIDA')
(categoria_vendas)

```

```python
 # do maior para o menor
categoria_vendas = categoria_vendas.sort_values(by='QUANTIDADE_VENDIDA', ascending=False)
(categoria_vendas)
```

```python
classe_compras = compras_unicas.groupby('CL_SEG').size().reset_index(name='TOTAL_COMPRAS')
(classe_compras)
```
